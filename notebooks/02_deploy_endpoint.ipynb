```python
import yaml

model_dir = "./knn_model_v2/knn-maintenance-notebook/model"

with open(f"{model_dir}/conda.yaml", "r") as f:
    conda_data = yaml.safe_load(f)

conda_data["dependencies"][-1]["pip"].extend([
    "azureml-inference-server-http",
    "azureml-ai-monitoring"
])

with open(f"{model_dir}/conda.yaml", "w") as f:
    yaml.safe_dump(conda_data, f, sort_keys=False)

print("conda.yaml updated")
```
